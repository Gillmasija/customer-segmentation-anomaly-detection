```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from google.colab import files
```

```python
file=files.upload()
```

```python
df=pd.read_csv("customers.csv")
df.head()
```

```python
df=df.drop(columns=["CustomerID","cluster"],axis=1)
```

```python
#selecting the columns for clustering
x=df[["Annual Income (k$)","Spending Score (1-100)"]]
```

```python
#load kmeans
kmeans = KMeans(n_clusters=3, random_state=42)
```

```python
#segmentation-assign each customer a segment
df["Segment"] = kmeans.fit_predict(x)
df.head()
```

```python
#display the segmentation
plt.scatter(x["Annual Income (k$)"],x["Spending Score (1-100)"],c=df["Segment"])
plt.xlabel("Annual Income (k$)")
plt.ylabel("Spending Score (1-100)")
plt.title("Customer segmentation")
plt.legend()
plt.show()
```

#Anomaly detection

```python
#cluster centers
centroids=kmeans.cluster_centers_
#calculate the distance of each point from the center
df["DistanceToCenter"]=df.apply(lambda x: np.linalg.norm(np.array([x["Annual Income (k$)"],x["Spending Score (1-100)"]])-centroids[int(x["Segment"])]),axis=1)
#anomaly threshold
threshold=(df["DistanceToCenter"].mean()+2*df["DistanceToCenter"].std())
df["Anomaly"]=df["DistanceToCenter"]>threshold
# Display detected anomalies
df.head()
```

```python
#display the output
plt.scatter(df["Annual Income (k$)"],df["Spending Score (1-100)"],c=df["Segment"],cmap="viridis")
#highlight the anomalies
plt.scatter(df[df["Anomaly"]==True]["Annual Income (k$)"],df[df["Anomaly"]==True]["Spending Score (1-100)"],color="red",s=100,label="Anomalies")
plt.xlabel("Annual Income")
plt.ylabel("Spending Score")
plt.title("Detected Anomalies")
plt.legend()
plt.show()
```
